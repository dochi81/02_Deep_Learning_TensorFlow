```python
import tensorflow as tf
```

```python
print("인식된 장치 목록:", tf.config.list_physical_devices())
```

```python
print("GPU 가속 최종 결과:", len(tf.config.list_physical_devices('GPU')) > 0)
# 사용 가능한 GPU 목록 확인
gpus = tf.config.list_physical_devices('GPU')
print(gpus)
```

```python
if gpus:
    try:
        for gpu in gpus:
            # 메모리 증가 옵션 켜기
            tf.config.experimental.set_memory_growth(gpu, True)
           
    except RuntimeError as e:
        print(e)


```
