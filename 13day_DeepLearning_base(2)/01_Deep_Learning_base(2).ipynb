```python
### 텐서플로우 프레임 워크
import tensorflow as tf

### 딥러닝 관련 라이브러리(모델 포함)
from tensorflow import keras

import numpy as np

from sklearn .model_selection import train_test_split

### 사용할 라이브러리 불러들이기
# 데이터 조작을 위한 라이브러리

import pandas as pd

# 기본 시각화 라이브러리

import matplotlib.pyplot as plt

# 통계적 시각화 라이브러리(보통 기본 시각화보다 향상된 라이브러리라고 생각하면 됨)

import seaborn as sns

# 시각화 시 그래프의 너비/높이를 기본 설정하기

#plt.rcParams["figure.figsize"] = (12, 5)

# 시각화시 한글 깨짐 처리를 위한 폰트 지정 (Window OS 기준)- 항상

#plt.rcParams["font.family"] = "Malgun Gothic"


# 시각화시 마이너스 (-) 기호 깨짐 방지 처리-항상

plt.rcParams["axes.unicode_minus"] =False

```

```python
""" 
1. 사용 데이터: fashion_mnist
2. 데이터 불러들이기
3. 데이터 스케일링
4. 2차원 데이터로 차원축소하기
5. 훈련: 검증 8:2로 분리하기

** 변수는 어제 사용한 변수명으로 
"""
```

```python
# 1~2. Fashion MNIST 데이터 불러오기
(train_input, train_target), (test_input, test_target) = (
    keras.datasets.fashion_mnist.load_data()
)

# 3. 데이터 스케일링 (0~255 → 0~1)
train_scaled = train_input / 255.0
test_scaled = test_input / 255.0

# 4. 2차원 데이터로 변환 (이미지마다 28×28 → 784)
train_scaled_2 = train_scaled.reshape(-1, 784)
test_scaled_2 = test_scaled.reshape(-1, 784)

# 5. 훈련 : 검증 = 8 : 2로 분리
train_scaled_2, val_scaled, train_target, val_target = train_test_split(
    train_scaled_2,
    train_target,
    test_size=0.2,
    random_state=42
)

print("훈련:", train_scaled_2.shape, train_target.shape)
print("검증:", val_scaled.shape, val_target.shape)
print("테스트:", test_scaled_2.shape, test_target.shape)
```

```python
### 픽셀값을 0~255에서 0~1 범위로 스케일링
train_scaled_2 = train_input / 255.0
test_scaled_2 = test_input / 255.0
```

### 신경망 모델에 계층 (Layer)추가하는 방법 (3가지)

```python
"""
1. 계층(Layer)을 먼저 만들고, 신경망 모델 생성시 추가하기
2. 신경망 모델 생성 시에, 계층 (Layer)를 함께 생성하여 추가하는 방식
3. 신경망 모델을 먼저 생성해 놓은 후,
add()함수를 이용하여 필요시 계층 (Layer)추가하는 방식 (주로 사용됨)
"""
```

### 1.계층 (Layer)을 먼저 만들고, 신경망 모델 만들기

```python
### Dense 계층 생성하기
# - 출력갯수 :100개
# - 활성화 함수: sigmoid 사용
# - 입력 특성 갯수: 784
# - 계층 변수명: dense1

dense1 = keras.layers.Dense (
    units =100,
    kernel_initializer="uniform",
    activation="sigmoid",
    input_shape=(784,)
)
dense1

### 반환(출력) 갯수 정의(다음 계층으로 전달되는 특성의 갯수가 됩니다.)
    # - 100의 의미 : 총784개 중에 주성분(주된 특성)의 특성으로 줄이는 갯수를 의미 - 차원을 줄여나감 (뜰채라고 생각하면 됨) = 하이퍼파라메터 = unit
    # - 숫자는 보통 메모리 사이즈를 기준으로 숫자값을 많이 사용합니다.(정해진 갯수는 없음)
    #   (1024, 512, 256, 128, 64, 32, 16, 8)
    #   : 딥러닝에서 주로 사용되는 숫자값 512~32 사이값 주로 사용됨
    #     (이외, 50, 100, 150, 200, 500, 1000의 값을 사용하는 경우도 있음)
```

```python
### 출력계층 (마지막 계층)생성하기
#- 출력계층 변수명: dense2

dense2 = keras.layers.Dense(
    units =10,

    ### softmax :다중 분류(종속변수의 범주의 갯수가 2개 인상, 보통 4개 이상에서 사용)
    # - 이진분류에서도 사용은 가능하지만 주로 이진분류는 sigmoid를 사용
    # - softmax는 범주의 갯수에 대한 각 확률값으로 추출

    activation="softmax",
)
dense2

```

```python
### 모델 생성과 동시에 위에 상성한 계층 추가하기
model =keras.Sequential([dense1,dense2])
model
```

```python
### 모델 계층확인
model.summary()
```

### 2.신경망 모델 생성시에 계층 (Layer)를 함께 생성하여 추가하는 방식

```python
### 모델 생성
model = keras.Sequential([
    ### 입력계층
    keras.layers.Dense(
        units=100, activation="sigmoid",
        input_shape=(784,), name="Input-Layer",
    ),

    ### 출력계층
    keras.layers.Dense(units=10, activation="softmax",
                        name="Output-Layer")
])

model
```

```python
model.summary()
```

### 3. 신경망 모델을 먼저 생성해 놓은 후, add()함수를 이용하여 필요시 계층 (Layer)추가하는 방식(주로 사용됨)

```python
### 모델 생성하기
model = keras.Sequential(name="Model")
model
```

```python
### 입력계층 추가하기
model.add(
    keras.layers.Dense(units=100,activation="sigmoid",
                       input_shape=(784,),name="input_Layer")

)
```

```python
### 출력계층 추가하기
model.add(
    keras.layers.Dense(units=10,activation="softmax",
                       name="Output_Layer")
)
```

### 모델 환경 설정하기 (compile)

```python
model.compile(
    ### 손실함수(10개를 기준으로 실행함)
    loss ="sparse_categorical_crossentropy",

    ### 훈련 중에 표시할 값(주로 정확도만 출력)
    metrics = ["accuracy"]
)
```

#### 모델 훈련시키기(fit)

```python
### 훈련 반복횟수(epochs) : 50회
### 훈련 시킬때마다 정확도 올라가고 손실율이 떨어지는데 모델 생성부터 다시 
### 실행해야하므로 모델생성부터는 한 코드셀에 작성하는 것이 편하다.
## 딥러닝 결과는 정확도보단 손실율로 결정됨 대부분

model.fit(train_scaled_2, train_target, epochs=50)
#model.fit(train_scaled, train_target, epochs=200)
```

### 훈련 모델 성능 검증 (평가)하기

### 성능향상 - 계층 추가

```python
#### 성능향상 - 은닉계층(Hidden-Layer)추가
```

```python
### 은닉계층 : 입력계층과 출력계층 중간에 포함되는 계층들을 의미함
```

```python
### 모델 생성

model = keras.Sequential(name ="Model")
model
```

```python
### 입력계층 (input Layer) 추가하기
# - 전처리 계층 사용 (차원 축소 계층 , 2차원을 1차원으로)
# - 전처리 계층은 성능에 영향을 미치지는 않음

# - Flatten() : 전처리 계층으로 2차원을 1차원으로 처리해주는 계층(많이 사용됨)
#             : 전처리 계층은 훈련 성능에 영향을 미치지 않습니다.
#             : 보통 Flatten() 전처리 계층은 "최초 입력계층으로 주로 사용"되며,
#               은닉계층으로도 사용될 수 있음(이후 CNN시에 은닉계층으로도 사용됨)
#             : 사용되는 데이터 (훈련/검증/테스트)들은 별도로 (너비 *높이)으로 만들지 않고,
#               원본 차원을 그대로 훈련에 넣어주면 전처리 계층이 알아서 차원을 축소함 


model.add(
    keras.layers.Flatten(
        ### 너비, 높이 값은 알고 있어야 함 
        input_shape=(28,28),
        name="Flatten-input-Layer"
    )
)
```

```python
### 은닉계층 (Hidden Layer)추가하기
# - 출력 갯수: 100개
# - 입력 계층 다음에 추가되는 계층을 은닉계층이라고 칭함 (출력계층이 뒤에 오게됨)

model.add(
    keras.layers.Dense(
        units=100,
        activation="relu",
        name="Hidden-Layer"

    )

)
```

```python
### 출력계층(Out Layer) 추가하기
model.add(
    keras.layers.Dense(
        units=10,
        activation="softmax",
        name="Output-Layer"
    )
)
```

```python
### 모델계층구조 확인하기
model.summary()
```

```python
"""
현재까지의 모델 계층구조를 사용해서
훈련 50회시키기...
- Flatten 전처리 계층에는 넣어주는 데이터는 2차원으로 차원 축소 이전의 3차원 원본 차원을 그대로 사용해야함 
- 원본 데이터를 다시 스케일링 및 훈련:검증 -8:2로 분리하여 데이터를 다시 구성하여 훈련 시켜야함
"""
```

```python
(train_input,train_target),(test_input,test_target) = keras.datasets.fashion_mnist.load_data()

train_scaled = train_input / 255.0
test_scaled = test_input / 255.0

train_scaled, val_scaled, train_target, val_target = train_test_split(train_scaled,train_target,test_size=0.2,random_state=42)

print(train_scaled.shape,train_target.shape)
print(val_scaled.shape,val_target.shape)
print(test_scaled_2.shape,test_target.shape)


model.compile(
    ### 손실함수(10개를 기준으로 실행함)
    loss ="sparse_categorical_crossentropy",

    ### 훈련 중에 표시할 값(주로 정확도만 출력)
    metrics = ["accuracy"]
)

model.fit(train_scaled, train_target, epochs=50)
```

```python
### 검증 및 테스트 데이터로 성능 검증(평가)하기

val_score = model.evaluate(val_scaled,val_target)
print(f"손실률:{val_score[0]:.5f},정확도:{val_score[1]:.5f}")

test_score = model.evaluate(test_scaled,test_target)
print(f"손실률:{test_score[0]:.5f}, 정확도:{test_score[1]:.5f}")
```

```python
### 성능향상- 옵티마이저 (Optimizer)는 모델 환경 설정 (compile)시에 함께 사용됨
```

```python
"""
<옵티마이저(Optimizer, 최적화) 기법>
 - 손실을 줄여나가기 위한 최적화(Optimizer) 방법을 의미함
 - 손실을 줄여나가는 최적화 방법으로 "경사하강법" 이론이 적용됩니다.
 - 특성들의 랜덤 시작 위치에서 목적지(종속변수 위치)까지 도달하기 위한
   기울어진 방향을 찾기 위한 방법을 의미합니다.
 - "경사하강법" 이론을 적용한 여러가지 방법들중 하나를 선택하여 정의
 - 옵티마이저(최적화) 방법
   : SGD(확률적경사하강법) < Adagrad < RMSProp(디폴트) < Adam, 이외 등등
 - 옵티마이저 설정 위치 : model.compile(optimizer="옵티마이저(최적화) 방법중 1개")
 
<옵티마이저(최적화) 방법 정의>
 * SGD(Stochastic Gradient Descent; 확률적 경사하강법)
  - 특성이 현재 위치에서 목적지까지 도달하는 과정 중에 보폭을 크게하여
    많은 길을 거치면서(극단적으로 방향을 바꿉니다.) 빠르게 탐색
  - 지그 재그 모양으로 탐색하면서 나아가는 방법
  - 아래 옵티마이저 방법들은 SGD를 근간으로 향상된 방법들 입니다.
  - 단점 : 보폭을 크게하면서 방향을 근단적으로 바꾸기 때문에
           주변을 정밀하게 확인하기 어려움
         : 보폭을 크게하기 때문에, 목적지를 건너 띄는 경우도 발생함
                                 (종속변수를 잘 못 찾는 경우 발생)
   
 * Adagrad
  - SGD의 큰 보폭에 대한 단점을 보완한 방법...
  - 학습률(보폭)을 적절하게 설정하기 위해 학습률 감소(보폭을 짧게)라는 기술 사용
  - 학습 진행 중에 학습률을 줄여가는 방법 사용
  - 처음에는 학습률(보폭)을 크게 학습하다가,
    점점 작게(보폭을 짧게) 학습한다는 개념을 적용
  - 이미 학습된 곳은 보폭을 크게하고, 학습이 완료되었던 곳은 보폭을 짧게 세밀하게 탐색
  - 손실이 더 이상 줄어들지 않으면(손실이 0이면) 종료
  - 단점, 목적지에 도달하지 않더라도 손실이 0이면 종료하는 단점이 있음
 
 * RMSprop(옵티마이저의 디폴트값 - 생략가능)
  - Adagrad는 학습량을 점점 작게 학습하기 때문에
    학습률(보폭)이 0이 되어 갱신되지 않는(학습되지 않는) 시점이 발생할 수 있는
    단점이 있음.
  - Adagrad의 단점을 보완하여,
    과거(이전)의 기울기 값을 반영하는 방식을 적용함
  - 먼 과거의 기울기(경사) 값은 조금만 반영하고,
    최근 기울기(경사)를 많이 반영하는 방식으로 처리됨
  - 과거 데이터를 저장해 놓아야 하기 때문에, 다소 훈련 시간이 걸림
  - 옵티마이저의 기본값(default)으로 사용됨, 생략가능
 
 * Adam
  - 공이 굴러가듯이 모멘텀(Momentum, 관성=방향담당)과 RMSProp을 융합한 방법
  - 방향(momentum)과 학습률(보폭)을 적절하게하여 탐색함
  - 자주 사용되는 기법으로, 좋은 결과를  얻을 수 있는 방법으로 유명함
 
 * Momentum(모멘텀)
  - 관성과 가속도를 적용하여 이동하던 방향으로 좀 더 유연하게 작동함
  (데이터는 급속하게 변화하지 않고, 점진적으로 변화함, 
  이를 관성과 가속도 이론을 적용하여 유연한 변화 흐름이 되도록 처리)
  - 메모리 사용이 많은 단점이 있음
    (과거 데이터를 저장해 놓고, 다음 과정에서 방향성(관성)을 이어 받아서 사용하게 됨)
"""




```

```python

model.compile(
    optimizer="sgd",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.fit(train_scaled,train_target,epochs=50)
```

```python
"""
<학습률(Learning Rate)>
 - 경사를 내려 올 때의 "보폭"이라고 이해
 - 옵티마이저 4개는 객체(클래스)로 되어 있습니다.
 - 클래스 생성 시에 학습률(learning_rate)을 정의 할 수 있습니다.
 - 학습률이 작을 수록 -> 보폭이 작음
           높을 수록  -> 보폭이 큼
 - 가장 손실이 작은 위치를 찾아서 움직이도록 작동됩니다.
 - 이때 가장 손실이 작은 위치는 모델이 스스로 찾아주기에 사람이 관여하지 않습니다.
 - 학습률(learning_rate) 값의 범위 : 0.1~0.0001 (기본값은 0.01, 생략가능)
 - 학습률의 값에 따라서 -> 훈련 성능에 영향을 미치기에
                           => 하이퍼파라메터 대상이 됩니다.

 - 과적합을 해소하기 위한 튜닝 방법으로 사용되기도 합니다.
   * 과대적합이 발생한 경우 : 학습률을 크게(보폭을 크게) 합니다.
   * 과소적합이 발생한 경우 : 학습률을 작게(보폭을 작게) 합니다.
"""
```

### 성능향상 - 옵티마이저에 학습율(보폭,learning rate)정의하기(하이퍼파라메터)

```python
"""
-1차원 전처리 계층을 입력 계층으로 사용^
- 은닉계층 1개 사용: 속성값은 자유롭게
- 옵티마이저 4개 모두 사용 ^
- 옵티마이저 학습율은 0.1,0.01 2개 각각 사용^
- 훈련 횟수 50회
- 성능 검증까지 진행
compile(run_early=True)
"""
```

```python
optimizer_classes = {
    "SGD": keras.optimizers.SGD,
    "Adagrad":keras.optimizers.Adagrad,
    "RMSprop":keras.optimizers.RMSprop,
    "Adam":keras.optimizers.Adam}

learning_rates=[0.1,0.01]

for name,optimizer_class in optimizer_classes.items():
    for lr in learning_rates:
       #print(f"\n옵티마이저: {name}, 학습률: {learning_rate}")

        model = keras.Sequential(name="model")

        model.add(keras.layers.Flatten(input_shape=(28,28)))
        model.add(keras.layers.Dense(100, activation="sigmoid"))
        model.add(keras.layers.Dense(10, activation="softmax"))
 
        model.compile(
        optimizer = optimizer_class(learning_rate=lr),
            loss = "sparse_categorical_crossentropy",
            metrics = ["accuracy"]
       )

        model.fit(train_scaled,train_target,epochs=50)  
           
                        
```

```python
### 검증 및 테스트 데이터로 성능 검증 (평가)하기

val_score = model.evaluate(val_scaled,val_target)
print(f"손실률:{val_score[0]:.5f},정확도:{val_score[1]:.5f}")

test_score = model.evaluate(test_scaled,test_target)
print(f"손실률:{test_score[0]:.5f}, 정확도:{test_score[1]:.5f}")
```
